**1. What is data ingestion and why is it important in data science? Which Python library is commonly used for data ingestion and cleaning?**

*   **Data Ingestion**: Refers to the process of obtaining and importing raw data from various disparate sources into a system where it can be stored, processed, and analyzed. This can involve pulling data from databases (SQL, NoSQL), APIs, log files, sensor data, web scraping, streaming data sources, or local files like CSVs, Excel sheets, and JSON files. It's the initial step that makes data available for any further operations.
*   **Importance in Data Science**: Data ingestion is paramount because without it, no data science activities can commence. It's the pipeline that feeds the entire data science workflow, from exploratory data analysis to model training and deployment. Effective ingestion ensures that data is accessible, timely, and in a format suitable for the subsequent cleaning and transformation stages.
*   **Common Python library**: The most widely used Python library for both data ingestion (especially from files and structured data sources) and cleaning is **Pandas**. It provides powerful and flexible data structures (like DataFrames) and functions (`read_csv`, `read_sql`, `fillna`, `dropna`, `apply`, etc.) that make these tasks efficient and straightforward.

**2. Why is data cleaning necessary before performing analysis or machine learning?**

Data cleaning (also known as data scrubbing or data wrangling) is a critical preprocessing step because real-world data is inherently imperfect. It often contains a myriad of issues such as:

*   **Missing Values**: Incomplete entries in columns.
*   **Inconsistent Formatting**: Dates in different formats, text with varying capitalization.
*   **Errors and Typos**: Human entry errors, measurement inaccuracies.
*   **Outliers**: Data points significantly different from others, potentially due to errors.
*   **Duplicates**: Redundant entries that can skew analysis.
*   **Irrelevant Data**: Features that do not contribute to the analytical goal.

If data cleaning is skipped, the downstream analysis and machine learning models will be built on flawed data, leading to:

*   **Inaccurate Insights**: Conclusions drawn from dirty data can be misleading or completely wrong.
*   **Poor Model Performance**: Machine learning models trained on noisy or inconsistent data will struggle to learn patterns effectively, resulting in low accuracy, poor generalization, and unreliable predictions.
*   **Biased Results**: Systematic errors or biases in the raw data can be amplified by models.
*   **Difficult Debugging**: It becomes very hard to diagnose why a model is performing poorly if the underlying data quality is unknown.

In essence, data cleaning ensures the integrity, consistency, and reliability of the data, which is fundamental for generating trustworthy results and building robust models.

**3. How do you load a CSV file into a Pandas DataFrame?**

Loading a CSV (Comma Separated Values) file into a Pandas DataFrame is a very common and simple operation. You use the `pd.read_csv()` function from the Pandas library. Here's a typical example:

```python
import pandas as pd

# Path to your CSV file
file_path = 'my_data.csv'

# Load the CSV file into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to confirm it loaded correctly
print(df.head())
```

The `pd.read_csv()` function is highly versatile and accepts many parameters to handle different CSV formats, such as specifying delimiters (`sep`), handling headers (`header`), parsing dates (`parse_dates`), and more.

**4. What is the difference between head() and tail() functions in Pandas?**

Both `df.head()` and `df.tail()` are convenient methods provided by Pandas DataFrames to quickly inspect the data. They are primarily used for preliminary data exploration and verification.

*   **`df.head(n=5)`**: This method returns the **first `n` rows** of the DataFrame. By default, if no number `n` is specified, it will return the top 5 rows. It's useful for checking the column names, data types, and initial values to ensure the data loaded correctly and looks as expected.
*   **`df.tail(n=5)`**: Conversely, this method returns the **last `n` rows** of the DataFrame. Similar to `head()`, it returns the bottom 5 rows by default. `tail()` is particularly useful for checking if all data has been loaded (especially if it's ordered by time) and for identifying any unusual entries or patterns at the end of the dataset.

In summary, `head()` gives you a glimpse of the beginning of your data, while `tail()` shows you the end, together offering a quick visual check of the DataFrame's overall structure and content.

````python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('/content/job_salary_prediction_dataset.csv')

print("Original Dataset:")
print(df.head())

# -------------------------------------------------
# 6. Remove Duplicate Rows
# -------------------------------------------------
print("\n--- Removing Duplicates ---")
df = df.drop_duplicates()
print("Duplicates removed!")

# -------------------------------------------------
# 7. Handle Missing Values (Mean / Median)
# -------------------------------------------------
print("\n--- Handling Missing Values ---")

# Select numeric columns
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Fill missing values with mean
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# If you want median instead, use below line:
# df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

print("Missing values handled!")

# -------------------------------------------------
# 8. Change Data Types of Selected Columns
# -------------------------------------------------
print("\n--- Changing Data Types ---")

# Example: Convert first numeric column to float (safe approach)
if len(numeric_cols) > 0:
    df[numeric_cols[0]] = df[numeric_cols[0]].astype(float)

print("Data types updated!")
print(df.dtypes)

# -------------------------------------------------
# 9. Filter Rows Based on Condition
# -------------------------------------------------
print("\n--- Filtering Data ---")

# Example: filter rows where first numeric column > its mean
if len(numeric_cols) > 0:
    filtered_df = df[df[numeric_cols[0]] > df[numeric_cols[0]].mean()]
    print("Filtered Data:")
    print(filtered_df.head())
else:
    print("No numeric column available for filtering.")

# -------------------------------------------------
# 10. Save Cleaned Dataset to New CSV
# -------------------------------------------------
print("\n--- Saving Cleaned Dataset ---")

df.to_csv('/content/cleaned_job_salary_dataset.csv', index=False)

print("Cleaned dataset saved successfully as 'cleaned_job_salary_dataset.csv'")
````
