## Data Wrangling and Pandas Concepts

### 1. What is data wrangling, and why is it important before analysis?

**Data wrangling**, also known as data munging or data preprocessing, is the process of cleaning, transforming, and organizing raw data into a more suitable format for analysis. It involves a series of critical steps to prepare data for downstream tasks like machine learning, statistical modeling, or visualization.

**Why it's important before analysis:**

*   **Ensures Accuracy and Reliability:** Raw data is often imperfect, containing errors, inconsistencies, and missing values. Data wrangling corrects these issues, ensuring that the insights derived from analysis are accurate and reliable. Faulty data leads to faulty conclusions.
*   **Improves Data Quality:** By standardizing formats, correcting inconsistencies, and handling outliers, data wrangling significantly enhances the overall quality of the dataset. High-quality data is foundational for robust analysis.
*   **Increases Efficiency:** Clean and well-structured data allows analysts and algorithms to work more efficiently. Time spent on wrangling upfront reduces time and effort during the analysis phase, preventing repetitive error handling.
*   **Facilitates Integration:** Often, data for analysis comes from multiple sources in various formats. Data wrangling helps integrate these disparate datasets into a unified, coherent structure, making them usable together.
*   **Prepares Data for Specific Tools/Models:** Different analytical tools and machine learning models have specific input requirements. Wrangling ensures data conforms to these requirements, making it compatible and preventing errors during model training or execution.
*   **Identifies Patterns and Anomalies:** The wrangling process itself, especially during exploratory phases, can help uncover initial patterns, trends, or anomalies that might be hidden in messy data.

### 2. You receive messy data with missing values, duplicates, and wrong types. What steps will you follow to clean it?

When confronted with messy data exhibiting missing values, duplicates, and incorrect data types, I would follow a systematic cleaning process:

1.  **Initial Data Inspection (Understanding the Mess):**
    *   **Load Data:** Load the data into a pandas DataFrame (`pd.read_csv()`, `pd.read_excel()`, etc.).
    *   **Quick Overview:** Use `df.head()`, `df.tail()`, `df.sample()` to get a quick glance at the data structure.
    *   **Check Data Types and Non-Null Counts:** `df.info()` is invaluable here. It shows column names, the number of non-null entries, and the inferred data types. This immediately highlights columns with missing values and potentially incorrect types (e.g., numbers stored as 'object' due to special characters).
    *   **Identify Column Names:** `df.columns` to see if names need standardization (e.g., removing spaces, special characters, making them lowercase).

2.  **Handling Missing Values:**
    *   **Identify Missing Values:** Use `df.isnull().sum()` to get a count of missing values per column. `df.isnull().sum() / len(df) * 100` gives the percentage.
    *   **Analyze Patterns:** Visualize missing data with `sns.heatmap(df.isnull(), cbar=False)` to understand if missingness is random or has a pattern.
    *   **Decide on Strategy (per column):**
        *   **Dropping:** If a column has a very high percentage of missing values (e.g., >70-80%) and is not critical, or if specific rows have too many missing values, drop them (`df.dropna(axis=1)` for columns, `df.dropna(axis=0)` for rows).
        *   **Imputation:**
            *   **Numerical:** Replace with mean, median (`df.fillna(df['col'].mean())`). Median is more robust to outliers.
            *   **Categorical:** Replace with mode (`df.fillna(df['col'].mode()[0])`).
            *   **Forward/Backward Fill:** For time-series data, `df.fillna(method='ffill')` or `df.fillna(method='bfill')` can be appropriate.
            *   **Advanced Imputation:** Use machine learning models (e.g., KNN Imputer) for more sophisticated imputation, if justified by the project.
    *   **Address Placeholders:** Replace non-standard missing value representations (e.g., '?', 'N/A', '-') with `np.nan` so pandas can recognize them (`df.replace(['?', 'N/A'], np.nan)`).

3.  **Addressing Duplicate Data:**
    *   **Identify Duplicates:** `df.duplicated().sum()` to count duplicate rows. `df[df.duplicated(keep=False)]` to view all duplicate rows.
    *   **Remove Duplicates:** `df.drop_duplicates(inplace=True)` to remove them, keeping the first occurrence by default. Consider `subset` parameter if duplicates should only be identified based on a specific set of columns.

4.  **Correcting Data Types:**
    *   **Review `df.info()`:** Identify columns with incorrect dtypes (e.g., numbers as `object`, dates as `object`).
    *   **Convert Types:**
        *   **Numerical:** Convert strings to numeric using `pd.to_numeric(df['col'], errors='coerce')`. `errors='coerce'` will turn unparseable values into `NaN`.
        *   **Dates:** Convert strings to datetime objects using `pd.to_datetime(df['col'], errors='coerce', format='%Y-%m-%d')`. Specifying a `format` can speed up conversion.
        *   **Categorical:** Convert object columns with a limited number of unique values to `category` dtype (`df['col'].astype('category']`) for memory efficiency and appropriate categorical operations.
        *   **Booleans:** Convert 'True'/'False' strings to actual boolean type.
    *   **Downcasting:** For memory optimization, downcast numerical types to their smallest possible representation (e.g., `int64` to `int32`, `float64` to `float32`). This can be done using `pd.to_numeric(df['col'], downcast='integer')`.

5.  **Handling Outliers (Post-Basic Cleaning):**
    *   **Detection:**
        *   **Visualization:** Box plots (`sns.boxplot()`), histograms (`sns.histplot()`), scatter plots are effective.
        *   **Statistical Methods:** Z-score for normally distributed data, IQR method (values outside `Q1 - 1.5*IQR` and `Q3 + 1.5*IQR`) for skewed data.
    *   **Treatment:**
        *   **Removal:** If outliers are clear errors and few, they can be removed.
        *   **Transformation:** Log transformation or square root transformation can reduce the impact of skewed distributions and outliers.
        *   **Capping/Winsorization:** Replace outliers with a specified percentile value (e.g., values above 99th percentile are set to the 99th percentile value).
        *   **Treat as Missing:** Convert outliers to `NaN` and then impute them.

6.  **Data Standardization/Normalization (if required for modeling):**
    *   Scaling numerical features to a common range (e.g., Min-Max Scaling) or standardizing them (e.g., Z-score normalization) for algorithms sensitive to feature scales.

7.  **Verification and Documentation:**
    *   After each major step, re-check `df.info()`, `df.describe()`, and `df.isnull().sum()` to confirm changes. Perform sanity checks.
    *   Document all cleaning steps taken, including the rationale, for reproducibility and transparency.

### 3. How can you check column names, data types, and summary statistics? What does "info()" show that "describe()" does not?

*   **Checking Column Names:**
    ```python
    df.columns
    # Output: Index(['col1', 'col2', 'col3'], dtype='object')
    ```
    This returns a pandas Index object containing all column labels.

*   **Checking Data Types and Non-Null Counts:**
    ```python
    df.info()
    ```
    This method prints a concise summary of a DataFrame, including:
    *   The index dtype and columns.
    *   The number of non-null values for each column.
    *   The `dtype` (data type) of each column.
    *   Memory usage.

*   **Checking Summary Statistics:**
    ```python
    df.describe()
    ```
    This method generates descriptive statistics that summarize the central tendency, dispersion, and shape of a dataset's distribution. By default, it works on numerical columns.
    *   For numerical data, it provides: `count`, `mean`, `std` (standard deviation), `min`, `25%` (1st quartile), `50%` (median), `75%` (3rd quartile), `max`.
    *   If `include='all'` is passed, it also includes statistics for object (categorical) data like `count`, `unique`, `top`, `freq`.

**What `df.info()` shows that `df.describe()` does not:**

*   **Non-Null Counts:** `info()` explicitly lists the number of non-null entries for each column, which is critical for quickly assessing the extent of missing data. `describe()` only provides the `count` of non-null values for numerical columns, not a direct comparison to the total rows.
*   **Exact Data Types (Dtype):** `info()` clearly displays the `dtype` for every column (e.g., `int64`, `float64`, `object`, `datetime64[ns]`). While `describe()` might implicitly show types through the statistics it calculates (or doesn't calculate for `object` types), `info()` gives a direct and complete overview.
*   **Total Entries:** `info()` states the total number of entries (rows) in the DataFrame.
*   **Memory Usage:** `info()` provides an estimate of the DataFrame's memory usage.

**What `df.describe()` shows that `df.info()` does not:**

*   **Statistical Measures:** `describe()` provides key statistical metrics like mean, standard deviation, min/max values, and quartiles, which are fundamental for understanding the distribution, spread, and range of numerical data.
*   **Categorical Statistics (with `include='all'`):** For categorical columns, `describe(include='all')` gives the number of unique values, the most frequent value (`top`), and its frequency (`freq`).

In essence, `info()` is for structural overview and data quality assessment (missing values, types), while `describe()` is for statistical summarization of the data's content.

### 4. How do you identify missing values and outliers in a dataset?

#### Identifying Missing Values:

1.  **Count Missing Values per Column:**
    ```python
    df.isnull().sum()
    # This gives a series where the index is the column name and the value is the count of NaN in that column.
    ```
    A similar approach can be used with `df.isna().sum()`.

2.  **Percentage of Missing Values per Column:**
    ```python
    df.isnull().sum() / len(df) * 100
    # This shows the percentage of missing values, which helps in prioritizing which columns to address.
    ```

3.  **Check for Any Missing Values:**
    ```python
    df.isnull().any().any()
    # Returns True if there is at least one missing value anywhere in the DataFrame, False otherwise.
    ```

4.  **Visualize Missing Data (Heatmap):**
    ```python
    import seaborn as sns
    import matplotlib.pyplot as plt

    plt.figure(figsize=(10, 6))
    sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
    plt.title('Missing Data Heatmap')
    plt.show()
    # White lines/blocks indicate missing values, giving a visual pattern of missingness.
    ```

5.  **Non-Standard Missing Values:** Sometimes missing values are not `NaN` but placeholders like '?', 'N/A', '-', 'None', or empty strings. These need to be identified manually (e.g., by inspecting unique values in object columns or using value counts) and then converted to `np.nan` for proper handling.
    ```python
    df['column'].value_counts() # To check for unusual entries
    df.replace(['?', 'N/A', 'None'], np.nan, inplace=True)
    ```

#### Identifying Outliers:

Outliers are data points that significantly deviate from other observations. Identification methods depend on the data distribution and context.

1.  **Visualization Methods (Most Common and Intuitive):**
    *   **Box Plots:** Excellent for numerical data. Outliers are typically shown as individual points beyond the 'whiskers' of the box plot. These whiskers usually extend 1.5 times the Interquartile Range (IQR) from the first (Q1) and third (Q3) quartiles.
        ```python
        sns.boxplot(x=df['numerical_column'])
        plt.title('Box Plot for Numerical Column')
        plt.show()
        ```
    *   **Histograms/KDE Plots:** Can reveal extreme values that are far from the main bulk of the distribution.
        ```python
        sns.histplot(df['numerical_column'], kde=True)
        plt.title('Histogram with KDE for Numerical Column')
        plt.show()
        ```
    *   **Scatter Plots:** Useful for identifying bivariate outliers (outliers in the relationship between two variables). Points far from the general cluster or trend are potential outliers.
        ```python
        sns.scatterplot(x='col_A', y='col_B', data=df)
        plt.title('Scatter Plot of Col A vs Col B')
        plt.show()
        ```

2.  **Statistical Methods:**
    *   **Z-score:** Measures how many standard deviations an element is from the mean. For data that is approximately normally distributed, values with a Z-score above a certain threshold (e.g., 2, 2.5, or 3) are often considered outliers.
        ```python
        from scipy.stats import zscore
        df['zscore'] = zscore(df['numerical_column'])
        outliers_zscore = df[abs(df['zscore']) > 3] # Example threshold of 3
        ```
    *   **IQR (Interquartile Range) Method:** More robust to skewed distributions than Z-scores. IQR = Q3 - Q1.
        *   Lower Bound: `Q1 - 1.5 * IQR`
        *   Upper Bound: `Q3 + 1.5 * IQR`
        Any data point outside these bounds is considered an outlier.
        ```python
        Q1 = df['numerical_column'].quantile(0.25)
        Q3 = df['numerical_column'].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        outliers_iqr = df[(df['numerical_column'] < lower_bound) | (df['numerical_column'] > upper_bound)]
        ```
    *   **Isolation Forest / DBSCAN:** For multi-dimensional data, more advanced machine learning algorithms can be employed for outlier detection.

### 5. How do you reduce memory usage in pandas DataFrames? Give long answer.

Reducing memory usage in pandas DataFrames is a critical optimization technique, especially when working with large datasets that might exceed available RAM. Efficient memory management can speed up operations and prevent out-of-memory errors. Here's a long answer detailing various strategies:

1.  **Use Appropriate Data Types (The Most Effective Method):**
    Pandas often defaults to `float64` for numerical data (even if it's integers with `NaN`s) and `int64` for integers. `object` type is used for strings and mixed types. These default types are often larger than necessary.

    *   **Integers (`int8`, `int16`, `int32`):** If an integer column's values fit within a smaller range, convert it. For example, if values are between -128 and 127, use `int8` instead of `int64` (which can store values up to ~9 quintillion). This can reduce memory by 8 times.
        ```python
        df['integer_col'] = df['integer_col'].astype('int32') # Or int16, int8
        ```
    *   **Floats (`float32`):** If high precision is not required, `float32` can often suffice, reducing memory by half compared to `float64`.
        ```python
        df['float_col'] = df['float_col'].astype('float32')
        ```
    *   **Categorical Data (`category` dtype):** This is one of the most significant memory savers for string columns with a limited number of unique values. The `category` dtype stores unique values as integers internally and maps them to their string representations, making it very efficient.
        ```python
        df['string_col'] = df['string_col'].astype('category')
        # Good for columns like 'gender', 'state', 'product_type' etc.
        ```
    *   **Booleans (`bool`):** Ensure boolean columns are stored as `bool` (1 byte) instead of `object` (which stores 'True'/'False' strings).
        ```python
        df['boolean_col'] = df['boolean_col'].astype('bool')
        ```
    *   **Datetime (`datetime64[ns]`):** While often memory-intensive, it's generally efficient for time-series data. However, ensure it's not stored as an `object` type. If only date or time is needed, you can extract components and store them as smaller integers or categories.

2.  **Downcasting Numerical Types (Automatic Optimization):**
    Pandas provides helper functions to automatically downcast numerical types to the smallest possible representation without losing information.

    ```python
    # Function to optimize numerical types
    def optimize_numerics(df):
        for col in df.select_dtypes(include=['int64', 'float64']).columns:
            if 'float' in str(df[col].dtype): # Handle floats first due to NaN
                if df[col].isnull().any(): # If NaNs present, cannot use integer types directly
                    df[col] = pd.to_numeric(df[col], downcast='float')
                else: # No NaNs, can attempt integer conversion if all values are whole numbers
                    # Check if all floats are actually integers without decimal part
                    if (df[col] == df[col].astype(int)).all():
                        df[col] = pd.to_numeric(df[col], downcast='integer')
                    else:
                        df[col] = pd.to_numeric(df[col], downcast='float')
            else: # Handle integers
                df[col] = pd.to_numeric(df[col], downcast='integer')
        return df

    # Function to optimize object types (mostly strings to category)
    def optimize_objects(df):
        for col in df.select_dtypes(include=['object']).columns:
            num_unique_values = len(df[col].unique())
            num_total_values = len(df[col])
            if num_unique_values / num_total_values < 0.5: # Heuristic: if unique values are less than 50% of total rows
                df[col] = df[col].astype('category')
        return df

    # Example usage:
    # print(df.info(memory_usage='deep')) # Before optimization
    # df = optimize_numerics(df)
    # df = optimize_objects(df)
    # print(df.info(memory_usage='deep')) # After optimization
    ```

3.  **Remove Unnecessary Columns:**
    If certain columns are not required for your analysis or model, drop them. This frees up memory occupied by those columns.
    ```python
    df.drop(columns=['unnecessary_col1', 'unnecessary_col2'], inplace=True)
    ```

4.  **Handle Missing Values Strategically:**
    *   `NaN` values in integer columns will force pandas to convert the column to `float64` because `NaN` is a float. If `NaN`s are not critical or can be filled with a suitable integer (e.g., -1, 0) or imputed before type conversion, you can then convert to a smaller integer type.
    *   For sparse data, consider using `pd.SparseDtype` or `pd.arrays.SparseArray` if a column contains a very high percentage of a single repeated value (often `NaN` or zero).

5.  **Use Efficient String Handling (if not using `category`):**
    *   For text data that is not suitable for `category` conversion (e.g., long, unique strings), if you need to store it, ensure you're not creating excessive temporary copies. Using string methods directly on pandas Series (`df['text_col'].str.lower()`) is generally efficient.
    *   If using regular expressions heavily, consider `re` module directly on series or compiling patterns.

6.  **Read Data in Chunks for Very Large Files:**
    If your dataset is too large to fit into memory even after optimization, read it in chunks. This allows you to process parts of the data iteratively or aggregate information without loading the entire file.
    ```python
    chunksize = 100_000
    df_list = []
    for chunk in pd.read_csv('large_data.csv', chunksize=chunksize):
        # Perform operations (e.g., cleaning, aggregation) on each chunk
        # chunk = optimize_numerics(chunk)
        # chunk = optimize_objects(chunk)
        df_list.append(chunk)
    df_final = pd.concat(df_list)
    ```

7.  **Delete Objects No Longer Needed:**
    After processing large intermediate DataFrames or variables, explicitly delete them and run garbage collection if memory is still an issue.
    ```python
    del large_intermediate_df
    import gc
    gc.collect()
    ```

8.  **Avoid Unnecessary Copies:**
    Operations like `df[df['col'] > 0]` create a *view* if possible, but many operations (e.g., `df.copy()`, `df.apply()`, `df.merge()`) create new DataFrames, temporarily doubling memory usage. Be mindful of when copies are made and only create them when necessary.

By systematically applying these techniques, you can significantly reduce the memory footprint of your pandas DataFrames, enabling you to work with larger datasets more effectively.
